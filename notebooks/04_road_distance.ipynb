# 4. Road distance to the nearest outlet

**Question.** How far do people in poorer and richer places live from their nearest outlet, by road (RQ4)?

**Method.**

- **People.** The 2024 census counts spread over WorldPop's 100 m pixels.
- **Network.** Every OpenStreetMap road of the classes counted in `road_km`, in both directions.
- **Joining.** Each pixel and each outlet is joined to its nearest road node by a straight leg.
- **Search.** One shortest-path search from all 7,284 placed outlets at once gives every node its distance to the nearest outlet.
- **Wealth groups.** The population-weighted RWI quintiles of notebook 1: by tile nationally, and by barangay among each city's barangays (NCR, Metro Cebu and Metro Davao).
- **No road route.** People on a road network with no outlet, mostly on small islands, are counted apart.

**Reading.** Nationally, distance falls steeply with wealth.

- **Poorest fifth.** The median person lives 11.5 km from an outlet by road, only 2% live within 2 km, and 4.4% have no road route to an outlet at all.
- **Richest fifth.** The median is 0.6 km, and 92% live within 2 km.

In NCR almost everyone (99%) lives within 2 km of an outlet, and the median of about 0.4 km is the same in every quintile.

The other two cities sit between NCR and the country:

- **Metro Cebu.** The poorest fifth's median is 4.5 km and only 24% of them live within 2 km. The other four fifths have medians of 0.6 to 1.0 km, and 82% to 97% live within 2 km.
- **Metro Davao.** The median falls from 12.5 km in the poorest fifth (1% within 2 km) to 4.1, 1.6, 1.0 and 0.7 km in the richest (95% within 2 km).

In both, the poorest fifth lives mostly in the outlying barangays: the hill barangays of Metro Cebu's interior, and Metro Davao's uplands and southern municipalities, where only 38% of the poorest fifth's people live in an urban cell.
So the distance gap there runs with the divide between the urban core and the hinterland as much as with wealth.
Straight-line distances are shorter, by about a quarter to a third nationally and in Metro Cebu and Metro Davao and by about a half in NCR, but tell the same story.
