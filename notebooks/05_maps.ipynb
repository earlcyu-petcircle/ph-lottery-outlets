# 5. Maps: outlet density beside wealth

**Question.** Where are outlets dense, where are people richer, and do the two line up on a map?
These are static versions of the interactive web map (`docs/map/`), saved to `results/figures/` so they can be read without a server; each code cell draws one map, and the image below it is the saved file.

**Classes.**

- **Outlet rate.** Outlets per 100,000 people, with "no outlet" as its own class. The units that have an outlet are split into fifths by their rate.
- **RWI.** Population-weighted fifths: each holds a fifth of the scope's people, and NCR is split by its own people.
- **Bivariate map.** It crosses wealth thirds of people with three outlet levels: none, at or below the scope's overall rate, and above it. These are the same classes and colours as the web map.
- **Grey.** Tiles or barangays with no people (or, for barangays, no RWI value).

A map shows where outlets and wealth sit together; it does not adjust for urban share, shops or province, which the models in notebooks 2 and 3 do.
